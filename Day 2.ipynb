```python
def isprime(num):
    for i in range(2,(num//2)+1):
        if num%i==0:
            print("Composite")
            break
        else:
            print("Prime")
            break
```

```python
isprime(33)
```

```python
for i in range(5):
    print(i, end = "--->")
    isprime(i)
```

```python
##### Ask the user for a number,print the next 5 prime numbers in the sequence
def isprime(n):
    for i in range(2,(n//2)+1):
        if n%i==0:
            return 0
    return 1

num = int(input("Enter a number: "))
count = 0
i=num+1
while count<5:
        if isprime(i):
            print(i)
            count+=1
            i+=1
        else: 
            i+=1
```
