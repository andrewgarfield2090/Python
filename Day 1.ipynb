# While Loop

```python
n=5
while(n<10):
    print(n)
    n+=1
print("We are now out of the loop")
```

```python
n=5
while(n<10):
    print(n)
    n-=1
    if n<0:
        break
```

```python
i=6
while True:
    a=int(input("Enter the number you wanna guess"))
    if(i==a):
        print("You've guessed it right")
        break
    if(i>a):
        print("Need more")
    if(i<a):
        print("Need less")
```

```python
# Q. three students ask for 3 subjects if (avg of 3)>75 print pass,else print pass  
```

```python
x=(int(input("Enter the no. of students: ")))
while x:
    a=(int(input("Enter the marks of subject A: ")))
    b=(int(input("Enter the marks of subject B: ")))
    c=(int(input("Enter the marks of subject C: ")))
    avg = (a+b+c)/3

    if avg>75:
        print('Pass')
    else:
        print('Fail')
    x-=1
   
```
