```python
l = [10,50,32,11,18,35]
```

```python
n = len(l)
for i in range(n):
    for j in range(n):
        if l[j]>l[i]:
            temp = l[i]
            l[i]=l[j]
            l[j]=temp
        else:
            continue
```

```python
def swap_num(l):
    n = len(l)
    for i in range(n):
        for j in range(n):
            if l[j]>l[i]:
                temp = l[i]
                l[i]=l[j]
                l[j]=temp
            else:
                continue
    return l
```

```python
def find_min(l):
    n = len(l)
    temp=99999
    for i in range(n):
        if l[i]<temp:
            temp = l[i]
        else:
            continue
    return temp
```

```python
l = [4,3,2,6,5,8,7,2]
find_min(l)
```
